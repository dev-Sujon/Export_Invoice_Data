```python
#### Clean the dataset
def clean_export_dataset(df):
    df = df.drop_duplicates().reset_index(drop=True)
    df.columns = df.columns.str.strip()
    df = df.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
    df = df.dropna(how='all').dropna(axis=1, how='all')
    return df

EIS = clean_export_dataset(EIS)
EIS.shape
```

```python
### Replace null value with blank

for col in EIS.columns:
    if EIS[col].dtype == "object":
        EIS[col] = EIS[col].fillna("")
    elif "datetime" in str(EIS[col].dtype):
        pass  # keep NaT
    else:
        EIS[col] = EIS[col].fillna(0)
EIS.sample(8)
```

```python
import pandas as pd

# Define columns by type
date_columns = [
    "Invoice_Date", "EXP_Date", "Ex_Factory_Date",
    "On_Board_Date", "Payment_Due_Date",
    "Shipping_Bill_Date", "BL_Date",
    "Bank_Submit_Date", "Realize_Date"
]

numeric_columns = [
    "LC/SC Value", "LCSC_Value", "LCSC_Qty", "Invoice_Qty",
    "Carton_Qty", "Invoice_Value", "Realize_Value"
]

text_columns = [
    "Factory", "Buyer", "Merchandiser", "Destination",
    "Payment_Term", "Bank", "Shipping_Mode", "Status",
    "Invoice_No", "EXP_No", "Ex_Factory_No", "Ex_Factory",
    "BL_No", "FDBC_No", "Realize_No"
]

boolean_columns = ["Is_FDBC", "Is_Realize"]

# Convert datatypes
for col in EIS.columns:
    if col in date_columns:
        EIS[col] = pd.to_datetime(EIS[col], errors="coerce")
    elif col in numeric_columns:
        EIS[col] = pd.to_numeric(EIS[col], errors="coerce")
    elif col in boolean_columns:
        EIS[col] = EIS[col].replace({"Yes": 1, "No": 0, "YES": 1, "NO": 0})
    elif col in text_columns:
        EIS[col] = EIS[col].astype(str).str.strip()
        # Fix scientific notation for ID columns
        if col in ["EXP_No", "Invoice_No", "Ex_Factory_No", "BL_No", "Realize_No"]:
            EIS[col] = EIS[col].str.replace(r"\.0$", "", regex=True)
    else:
        pass  # leave any other columns as-is

print(EIS.dtypes)
print("✅ Data types converted successfully.")
```

```python
import pandas as pd
import numpy as np

# =========================================
# COPY DATA & REMOVE DUPLICATE COLUMNS
# =========================================
df = EIS.copy()
df = df.loc[:, ~df.columns.duplicated()]

# =========================================
# CONVERT DATE COLUMNS
# =========================================
date_cols = [
    'Invoice_Date', 'EXP_Date', 'Ex_Factory_Date',
    'On_Board_Date', 'Bank_Submit_Date', 'Realize_Date'
]

for col in date_cols:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')

# =========================================
# CONVERT NUMERIC COLUMNS
# =========================================
df['Invoice_Value'] = pd.to_numeric(df['Invoice_Value'], errors='coerce')
df['PaymentAfterExFactory'] = pd.to_numeric(
    df['PaymentAfterExFactory'], errors='coerce'
)

# =========================================
# REMOVE EMPTY FDBC_No
# =========================================
df = df[df['FDBC_No'].notna()]

# =========================================
# GROUP BY FDBC_No
# =========================================
summary = df.groupby('FDBC_No', as_index=False).agg({

    # First match (XLOOKUP equivalent)
    'Status': 'first',
    'Category': 'first',
    'Buyer': 'first',
    'Factory': 'first',
    'Bank': 'first',
    'Payment_Term': 'first',
    'PaymentAfterExFactory': 'first',

    # Date logic
    'Ex_Factory_Date': 'max',
    'Bank_Submit_Date': 'max',

    # COUNTIF
    'Invoice_No': 'count',

    # SUMIF
    'Invoice_Value': 'sum'
})

# Rename columns properly
summary.rename(columns={
    'Factory': 'Company',
    'Payment_Term': 'PaymentTerm',
    'PaymentAfterExFactory': 'PaymentTermAfterExFactory',
    'Ex_Factory_Date': 'LastExFactoryDate',
    'Bank_Submit_Date': 'LastBankSubmitDate',
    'Invoice_No': 'TotalInvoice',
    'Invoice_Value': 'InvoiceValue'
}, inplace=True)

# =========================================
# IsBankSubmit?
# =========================================
summary['IsBankSubmit'] = np.where(
    summary['LastBankSubmitDate'].notna(),
    'YES',
    'NO'
)

# =========================================
# OnBoardDate Range (MIN + MAX)
# =========================================
onboard = (
    df.groupby('FDBC_No')['On_Board_Date']
      .agg(['min', 'max'])
      .reset_index()
)

summary = summary.merge(onboard, on='FDBC_No', how='left')

summary['OnBoardDate_Range'] = np.where(
    summary['min'].notna(),
    summary['min'].dt.strftime('%d-%b-%Y') +
    " to " +
    summary['max'].dt.strftime('%d-%b-%Y'),
    ""
)

summary.drop(columns=['min', 'max'], inplace=True)

# =========================================
# INVOICES (TEXTJOIN Equivalent)
# =========================================
invoice_list = (
    df.sort_values('Invoice_No')
      .drop_duplicates(['FDBC_No', 'Invoice_No'])
      .groupby('FDBC_No')['Invoice_No']
      .agg(','.join)
      .reset_index(name='INVOICES')
)

summary = summary.merge(invoice_list, on='FDBC_No', how='left')

# =========================================
# IsFDBCRealizable?
# =========================================
summary['PaymentTermAfterExFactory'] = pd.to_numeric(
    summary['PaymentTermAfterExFactory'],
    errors='coerce'
)

today = pd.Timestamp.today()

due_date = (
    summary['LastExFactoryDate'] +
    pd.to_timedelta(summary['PaymentTermAfterExFactory'], unit='D')
)

summary['IsFDBCRealizable'] = np.where(
    summary['PaymentTermAfterExFactory'].notna(),
    np.where(
        today > due_date,
        'NeedRealize',
        'NotYetRealize'
    ),
    ""
)

# =========================================
# FINAL TABLE
# =========================================
FDBC_Summary_Table = summary

print("FDBC Summary Table Created Successfully")
```
