# 📦 Export Invoice Summary — Pending Reports

**Purpose:** Load, clean, and transform Export Invoice Summary (EIS) data for Woven and Sweater categories, then generate four pending-status reports:
- **Realize Pending**
- **On Board Pending**
- **Ex-Factory Pending**
- **Bank Submit Pending**

## 6. Basic Cleaning

```python
def clean_export_dataset(df: pd.DataFrame) -> pd.DataFrame:
    """Remove duplicates, strip whitespace, drop all-null rows/columns."""
    df = df.drop_duplicates().reset_index(drop=True)
    df.columns = df.columns.str.strip()
    df = df.apply(lambda col: col.str.strip() if col.dtype == 'object' else col)
    df = df.dropna(how='all').dropna(axis=1, how='all')
    return df

EIS = clean_export_dataset(EIS)
print(f"After cleaning → {EIS.shape[0]:,} rows × {EIS.shape[1]} cols")
```

---
## 7. Data-Type Conversion

> **Bug fix — `Invoice_Date` NaT issue:**  
> The original notebook called `fillna("")` on all object columns *before* converting dates.
> This silently replaced genuine `NaN` / blank date cells with an empty string `""`,
> which `pd.to_datetime()` cannot parse → every row returned `NaT`.
>
> **Fix:** Convert date columns to `datetime64` **first**, then fill missing values for
> non-date columns only.

```python
# ─── Column Classification ─────────────────────────────────────────────────────
DATE_COLUMNS = [
    'Invoice_Date', 'EXP_Date', 'Ex_Factory_Date',
    'On_Board_Date', 'Payment_Due_Date',
    'Shipping_Bill_Date', 'BL_Date',
    'Bank_Submit_Date', 'Realize_Date'
]

NUMERIC_COLUMNS = [
    'LCSC_Value', 'LCSC_Qty', 'Invoice_Qty',
    'Carton_Qty', 'Invoice_Value', 'Realize_Value'
]

ID_COLUMNS = [
    'EXP_No', 'Invoice_No', 'Ex_Factory_No', 'BL_No', 'Realize_No'
]

TEXT_COLUMNS = [
    'Factory', 'Buyer', 'Merchandiser', 'Destination',
    'Payment_Term', 'Bank', 'Shipping_Mode', 'Status',
    'FDBC_No',
] + ID_COLUMNS

BOOLEAN_COLUMNS = ['Is_FDBC', 'Is_Realize']

# ─── STEP 1: Convert date columns FIRST (before any fillna) ───────────────────
# This is the critical fix: date columns must be converted BEFORE filling blanks,
# otherwise empty-string fills prevent correct datetime parsing → NaT everywhere.
for col in DATE_COLUMNS:
    if col in EIS.columns:
        # Replace empty strings with NaN so to_datetime can handle them
        EIS[col] = EIS[col].replace('', np.nan)
        EIS[col] = pd.to_datetime(EIS[col], dayfirst=False, errors='coerce')

# ─── STEP 2: Fill missing values for non-date columns ─────────────────────────
for col in EIS.columns:
    if col in DATE_COLUMNS:
        pass  # Leave NaT as-is — intentional for date columns
    elif col in NUMERIC_COLUMNS:
        EIS[col] = pd.to_numeric(EIS[col], errors='coerce').fillna(0)
    elif col in BOOLEAN_COLUMNS:
        EIS[col] = EIS[col].replace({'Yes': 1, 'YES': 1, 'No': 0, 'NO': 0})
    elif col in TEXT_COLUMNS:
        EIS[col] = EIS[col].astype(str).str.strip()
        if col in ID_COLUMNS:
            EIS[col] = EIS[col].str.replace(r'\.0$', '', regex=True)
        EIS[col] = EIS[col].replace('nan', '')
    else:
        if EIS[col].dtype == 'object':
            EIS[col] = EIS[col].fillna('')

print("✅ Data types converted successfully.")
print("\n📅 Date column dtypes (all should be datetime64):")
for col in DATE_COLUMNS:
    if col in EIS.columns:
        non_nat = EIS[col].notna().sum()
        print(f"  {col:<25} → {str(EIS[col].dtype):<20} | {non_nat:,} valid dates")
```

## 10. Realize Pending Report

```python
Realize_Report = EIS[EIS['Status'] == 'Realize Pending'].copy()
print(f"Realize Pending records: {Realize_Report.shape[0]:,}")
Realize_Report.sample(min(5, len(Realize_Report)))
```

```python
# ─── Build Realize Pending Summary ────────────────────────────────────────────
df = Realize_Report.copy()

# Normalize FDBC_No — treat 'nan' strings as missing
df['FDBC_No'] = df['FDBC_No'].astype(str).str.strip().replace('nan', np.nan)
df = df.dropna(subset=['FDBC_No']).copy()

invoice_col = 'Invoice_No'

# ─── OnBoard Date Range (min–max per FDBC) ────────────────────────────────────
onboard_range = (
    df.groupby('FDBC_No', as_index=False)['On_Board_Date']
      .agg(min_onboard='min', max_onboard='max')
)

# ─── Invoice List (unique, sorted) ────────────────────────────────────────────
invoice_list = (
    df.dropna(subset=[invoice_col])
      .assign(_inv=lambda d: d[invoice_col].astype(str).str.strip())
      .sort_values('_inv')
      .drop_duplicates(['FDBC_No', '_inv'])
      .groupby('FDBC_No', as_index=False)['_inv']
      .agg(INVOICES=lambda s: ', '.join(s))
)

# ─── Main Aggregation ─────────────────────────────────────────────────────────
summary = (
    df.groupby('FDBC_No', as_index=False)
      .agg(
          Status               = ('Status',               'first'),
          Category             = ('Category',             'first'),
          Buyer                = ('Buyer',                'first'),
          Company              = ('Factory',              'first'),
          Bank                 = ('Bank',                 'first'),
          PaymentTermXFactory  = ('PaymentAfterExFactory','first'),
          ExFactoryDate        = ('Ex_Factory_Date',      'max'),
          BankSubmitDate       = ('Bank_Submit_Date',     'max'),
          InvoiceDate          = ('Invoice_Date',         'max'),   # ← fixed: actual datetime
          Total_Invoices       = (invoice_col,            lambda s: s.dropna().count()),
          Total_Invoice_Value  = ('Invoice_Value',        'sum')
      )
)

# ─── Merge Supplementary Data ─────────────────────────────────────────────────
summary = summary.merge(onboard_range, on='FDBC_No', how='left')
summary = summary.merge(invoice_list,  on='FDBC_No', how='left')
summary['INVOICES'] = summary['INVOICES'].fillna('')

# ─── IsBankSubmit? ─────────────────────────────────────────────────────────────
summary['IsBankSubmit?'] = np.where(summary['BankSubmitDate'].notna(), 'YES', 'NO')

# ─── OnBoardDate(Range) ────────────────────────────────────────────────────────
def format_onboard(rmin, rmax):
    if pd.isna(rmin) or pd.isna(rmax):
        return ''
    if rmin == rmax:
        return rmin.strftime('%d-%b-%Y')
    return f"{rmin.strftime('%d-%b-%Y')} to {rmax.strftime('%d-%b-%Y')}"

summary['OnBoardDate(Range)'] = summary.apply(
    lambda r: format_onboard(r['min_onboard'], r['max_onboard']), axis=1
)
summary.drop(columns=['min_onboard', 'max_onboard'], inplace=True)

# ─── IsFDBCRealizable? ─────────────────────────────────────────────────────────
today = pd.Timestamp.today().normalize()
summary['PaymentTermXFactory'] = pd.to_numeric(summary['PaymentTermXFactory'], errors='coerce')

has_term_and_date = summary['PaymentTermXFactory'].notna() & summary['ExFactoryDate'].notna()
summary['Due_Date'] = pd.NaT
summary.loc[has_term_and_date, 'Due_Date'] = (
    summary.loc[has_term_and_date, 'ExFactoryDate']
    + pd.to_timedelta(summary.loc[has_term_and_date, 'PaymentTermXFactory'].astype(int), unit='D')
)

def realizable_status(row):
    if pd.isna(row['Due_Date']):
        return ''
    return 'NeedRealize' if today > row['Due_Date'] else 'NotYetRealize'

summary['IsFDBCRealizable?'] = summary.apply(realizable_status, axis=1)
summary.drop(columns=['Due_Date'], inplace=True)

# ─── Final Column Order ────────────────────────────────────────────────────────
summary.insert(0, 'SL#', range(1, len(summary) + 1))

REALIZE_FINAL_COLS = [
    'SL#', 'FDBC_No', 'Status', 'Category', 'Buyer', 'Company',
    'IsBankSubmit?', 'ExFactoryDate', 'Bank', 'BankSubmitDate',
    'PaymentTermXFactory', 'OnBoardDate(Range)', 'Total_Invoices',
    'Total_Invoice_Value', 'InvoiceDate', 'IsFDBCRealizable?', 'INVOICES'
]

for c in REALIZE_FINAL_COLS:
    if c not in summary.columns:
        summary[c] = ''

summary = summary[REALIZE_FINAL_COLS]
summary['Total_Invoice_Value'] = pd.to_numeric(summary['Total_Invoice_Value'], errors='coerce').fillna(0)

# ─── Exclude Known Invalid FDBCs ──────────────────────────────────────────────
EXCLUDE_FDBC = ['FDBC-278/24', '302/24', '0976/24', '0891/24', '0938/24', '040/24']
summary = summary[~summary['FDBC_No'].isin(EXCLUDE_FDBC)]

RealizePending = summary.reset_index(drop=True)
RealizePending['SL#'] = range(1, len(RealizePending) + 1)

print(f"✅ Realize Pending report created: {RealizePending.shape[0]:,} records")
print(f"   InvoiceDate valid count: {RealizePending['InvoiceDate'].notna().sum():,}")
```

## 13. Bank Submit Pending Report

```python
BANKSUBMIT_DROP_COLS = [
    'LCSC_Qty', 'Carton_Qty', 'Destination', 'Ex_Factory_No', 'Ex_Factory',
    'Shipping_Bill_Date', 'BL_Date', 'Realize_No', 'Realize_Date',
    'Realize_Value', 'PaymentAfterExFactory'
]

banksubmit_df = EIS[EIS['Status'] == 'Bank Submit Pending'].copy()
existing_drops = [c for c in BANKSUBMIT_DROP_COLS if c in banksubmit_df.columns]
banksubmit_df = banksubmit_df.drop(columns=existing_drops)
banksubmit_df.insert(0, 'SL#', range(1, len(banksubmit_df) + 1))

# Bug fix: original had 'Invoice_Date' listed twice in final_cols
BANKSUBMIT_FINAL_COLS = [
    'SL#', 'Status', 'LCSC_No', 'LCSC_Value', 'Factory', 'Buyer',
    'Ex_Factory', 'Ex_Factory_Date', 'Shipping_Mode',
    'FDBC_No', 'EXP_No', 'EXP_Date',
    'Invoice_No', 'Invoice_Date',              # ← de-duplicated (was listed twice)
    'Destination', 'Merchandiser',
    'On_Board_Date', 'Bank', 'Bank_Submit_Date',
    'Is_FDBC', 'Is_Realize'
]
for col in BANKSUBMIT_FINAL_COLS:
    if col not in banksubmit_df.columns:
        banksubmit_df[col] = ''

BankSubmitPending = banksubmit_df[BANKSUBMIT_FINAL_COLS].copy()

print(f"✅ Bank Submit Pending report created: {BankSubmitPending.shape[0]:,} records")
BankSubmitPending.head(5)
```

```python
# ─── Export All Pending Reports (Multi-Sheet) ─────────────────────────────────
date_str   = datetime.now().strftime("%d%m%Y")
report_path = OUTPUT_DIR / f"Pending_Report_{date_str}.xlsx"

sheets = {
    'RealizePending'    : RealizePending,
    'OnBoardPending'    : OnBoardPending,
    'ExFactoryPending'  : ExFactoryPending,
    'BankSubmitPending' : BankSubmitPending,
}

try:
    with pd.ExcelWriter(report_path, engine='xlsxwriter') as writer:
        for sheet_name, df_sheet in sheets.items():
            if df_sheet is not None and not df_sheet.empty:
                df_sheet.to_excel(writer, sheet_name=sheet_name, index=False)
                print(f"  ✔ {sheet_name:<22} → {df_sheet.shape[0]:>4} rows")
            else:
                print(f"  ⚠ {sheet_name} skipped (empty)")
    print(f"\n✅ Pending report saved → {report_path}")
except Exception as e:
    print(f"❌ Error saving report: {e}")
```

### Bugs Fixed in This Version
1. **`Invoice_Date` NaT** — Date columns are now converted to `datetime64` **before** the `fillna('')` loop runs on object columns. Previously, the fill replaced blank cells with `""` strings, which `pd.to_datetime()` cannot parse, causing every date to become `NaT`.
2. **`ExFactoryPending` missing comma** — `'Merchandiser'` and `'On_Board_Date'` were silently concatenated into a single invalid column name `'MerchandiserOn_Board_Date'`.
3. **`BankSubmitPending` duplicate column** — `'Invoice_Date'` appeared twice in the final column list, which raises a `ValueError` in pandas. De-duplicated.
4. **Defensive drops** — All `drop()` calls now filter to only existing columns, preventing `KeyError` on datasets that differ slightly from the expected schema.
