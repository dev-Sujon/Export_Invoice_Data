## 3. Database Connection

### 🔍 Bug #1 — Login Failed (Error 18456)

**Original (broken):**
```
Trusted_Connection=yes   ← uses Windows Authentication → MOHAMMADI\Sujon
```
**Fixed:**
```
UID=sa;PWD=*userdts*     ← uses SQL Server Authentication → sa
```

````python
def build_engine(server, database, user, password, driver):
    """
    Build a SQLAlchemy engine using SQL Server Authentication.

    NOTE: 'Trusted_Connection=yes' was REMOVED — it causes Windows Auth
    to override the UID/PWD and fails with Error 18456 on non-domain machines.
    """
    connection_string = (
        f"DRIVER={{{driver}}};"
        f"SERVER={server};"
        f"DATABASE={database};"
        f"UID={user};"
        f"PWD={password};"
        "Encrypt=yes;"
        "TrustServerCertificate=yes;"
        "Connection Timeout=30;"
    )
    params = urllib.parse.quote_plus(connection_string)
    engine = create_engine(
        f"mssql+pyodbc:///?odbc_connect={params}",
        fast_executemany=True,   # better batch performance
        pool_pre_ping=True,      # auto-reconnect on stale connections
    )
    return engine


engine = build_engine(DB_SERVER, DB_NAME, DB_USER, DB_PASSWORD, DB_DRIVER)
print(f"Engine created → {engine}")
````

````python
def test_connection(engine):
    """Ping the database and return server metadata."""
    try:
        with engine.connect() as conn:
            result = conn.execute(text("""
                SELECT
                    @@SERVERNAME    AS server_name,
                    DB_NAME()       AS database_name,
                    SYSTEM_USER     AS login_user,
                    @@VERSION       AS sql_version
            """)).fetchone()

        print("✅ Connection Successful!")
        print(f"   Server   : {result.server_name}")
        print(f"   Database : {result.database_name}")
        print(f"   User     : {result.login_user}")
        print(f"   Version  : {result.sql_version[:80]}...")
        return True

    except Exception as exc:
        print(f"❌ Connection Failed: {exc}")
        print("\n💡 Troubleshooting Tips:")
        print("   1. Verify SERVER address and PORT are correct.")
        print("   2. Ensure the firewall allows port 1437 outbound.")
        print("   3. Confirm DB_USER / DB_PASSWORD are correct SQL Server credentials.")
        print("   4. Make sure 'SQL Server Authentication' is enabled on the server.")
        return False


test_connection(engine)
````

## 4. Load Invoice Data

````python
def list_tables(engine, search: str = "") -> pd.DataFrame:
    """List all user tables in the database, optionally filtered by name."""
    query = text("""
        SELECT
            TABLE_SCHEMA  AS [Schema],
            TABLE_NAME    AS [Table],
            TABLE_TYPE    AS [Type]
        FROM INFORMATION_SCHEMA.TABLES
        WHERE TABLE_TYPE = 'BASE TABLE'
          AND TABLE_NAME LIKE :search
        ORDER BY TABLE_SCHEMA, TABLE_NAME
    """)
    with engine.connect() as conn:
        df = pd.read_sql(query, conn, params={"search": f"%{search}%"})
    return df


# Show all tables — change search= to filter (e.g. search="Invoice")
tables = list_tables(engine, search="")
print(f"Found {len(tables)} table(s) in '{DB_NAME}':")
tables
````

````python
def inspect_columns(engine, table_name: str, schema: str = "dbo") -> pd.DataFrame:
    """Return column names and data types for a given table."""
    query = text("""
        SELECT
            COLUMN_NAME                         AS [Column],
            DATA_TYPE                           AS [DataType],
            CHARACTER_MAXIMUM_LENGTH            AS [MaxLen],
            IS_NULLABLE                         AS [Nullable],
            COLUMN_DEFAULT                      AS [Default]
        FROM INFORMATION_SCHEMA.COLUMNS
        WHERE TABLE_SCHEMA = :schema
          AND TABLE_NAME   = :table
        ORDER BY ORDINAL_POSITION
    """)
    with engine.connect() as conn:
        df = pd.read_sql(query, conn, params={"schema": schema, "table": table_name})
    return df


# ─── UPDATE THIS: replace with your actual invoice table name ─────────────────
INVOICE_TABLE = "YourInvoiceTableName"   # e.g. "SalesInvoice", "tbl_Invoice"
# ─────────────────────────────────────────────────────────────────────────────

cols = inspect_columns(engine, INVOICE_TABLE)
print(f"Columns in [{INVOICE_TABLE}]:")
cols
````

````python
def load_invoices(
    engine,
    table: str,
    date_columns: list[str] | None = None,
    limit: int | None = None,
) -> pd.DataFrame:
    """
    Load invoice records from SQL Server.

    Parameters
    ----------
    engine       : SQLAlchemy engine
    table        : Table name in DBMGERP
    date_columns : List of date column names to parse explicitly (fixes NaT bug)
    limit        : Optional row cap for exploration

    Returns
    -------
    pd.DataFrame
    """
    top_clause = f"TOP {limit}" if limit else ""
    query = f"SELECT {top_clause} * FROM {table} WITH (NOLOCK)"

    print(f"⏳ Loading data from [{table}]...")
    with engine.connect() as conn:
        # ── FIX #2: Do NOT pass parse_dates here ─────────────────────────────
        # pandas' auto parse_dates can return NaT for SQL Server datetime
        # columns when the dtype comes through as 'object' or 'datetime64[ns]'
        # without timezone. We convert manually below.
        df = pd.read_sql(query, conn)

    print(f"✅ Loaded {len(df):,} rows × {len(df.columns)} columns.")
    return df


# Load with optional row limit for inspection
df_raw = load_invoices(engine, INVOICE_TABLE, limit=10_000)
df_raw.head()
````

## 5. Fix & Validate `Invoice Date`

### 🔍 Bug #2 — Invoice Date Shows `NaT`

**Why it happens:**  
SQL Server `datetime` columns sometimes arrive in pandas as Python `datetime` objects wrapped in an `object`-typed Series. pandas' auto-parsing (`parse_dates=`) fails silently and returns `NaT`.  

**Fix:**  
Explicitly convert every date column after loading using `pd.to_datetime(..., errors='coerce')`. `errors='coerce'` turns any genuinely unparseable values into `NaT` so the column stays clean.

````python
# ─── UPDATE THIS: list every date/datetime column in your table ───────────────
DATE_COLUMNS = [
    "InvoiceDate",       # ← adjust to match your actual column names
    # "DueDate",
    # "PaymentDate",
    # "CreatedAt",
]
# ─────────────────────────────────────────────────────────────────────────────

def fix_date_columns(df: pd.DataFrame, date_cols: list[str]) -> pd.DataFrame:
    """
    Explicitly parse date columns to datetime64[ns].

    Strategy:
      1. If the column is already datetime, strip timezone info (tz-aware →
         tz-naive) so downstream operations stay consistent.
      2. If the column is object/str, coerce with pd.to_datetime.
      3. Report any NaT values that remain (genuinely missing / unparseable).
    """
    df = df.copy()

    for col in date_cols:
        if col not in df.columns:
            print(f"⚠️  Column '{col}' not found — skipping.")
            continue

        before_nulls = df[col].isna().sum()

        # Step 1: try direct conversion
        df[col] = pd.to_datetime(df[col], errors="coerce", utc=False)

        # Step 2: strip timezone if present (SQL Server sometimes sends UTC-aware)
        if hasattr(df[col].dtype, "tz") and df[col].dtype.tz is not None:
            df[col] = df[col].dt.tz_convert(None)  # convert to local naive

        after_nulls = df[col].isna().sum()
        new_nats    = after_nulls - before_nulls

        status = "✅" if after_nulls == 0 else "⚠️ "
        print(
            f"{status} '{col}' → {df[col].dtype} | "
            f"nulls before: {before_nulls} | after: {after_nulls} | "
            f"new NaT introduced: {new_nats}"
        )

        if after_nulls > 0:
            sample = df.loc[df[col].isna(), col].head(3)
            print(f"    Sample raw values that failed: {list(sample)}")

    return df


df = fix_date_columns(df_raw, DATE_COLUMNS)
print("\nDone. Current dtypes:")
print(df[DATE_COLUMNS].dtypes)
````

````python
def validate_date_range(
    df: pd.DataFrame,
    col: str,
    min_year: int = 2000,
    max_year: int = 2030,
) -> None:
    """Report date range statistics and flag suspicious values."""
    series = df[col].dropna()
    if series.empty:
        print(f"⚠️  '{col}' has no valid dates.")
        return

    print(f"─── {col} ────────────────────────────────")
    print(f"  Total rows    : {len(df):,}")
    print(f"  Valid dates   : {len(series):,}")
    print(f"  Missing (NaT) : {df[col].isna().sum():,}")
    print(f"  Earliest      : {series.min()}")
    print(f"  Latest        : {series.max()}")

    out_of_range = df[
        (df[col].dt.year < min_year) | (df[col].dt.year > max_year)
    ]
    if not out_of_range.empty:
        print(f"  ⚠️  {len(out_of_range):,} records outside [{min_year}–{max_year}]:")
        print(out_of_range[col].value_counts().head(5))
    else:
        print(f"  ✅ All dates are within [{min_year}–{max_year}]")


for col in DATE_COLUMNS:
    if col in df.columns:
        validate_date_range(df, col)
````

## 6. Data Summary & Preview

````python
def data_overview(df: pd.DataFrame) -> None:
    """Print a concise quality summary of the DataFrame."""
    total_rows, total_cols = df.shape
    null_counts = df.isna().sum()
    null_pct    = (null_counts / total_rows * 100).round(1)

    summary = pd.DataFrame({
        "dtype":        df.dtypes,
        "non_null":     df.notna().sum(),
        "null_count":   null_counts,
        "null_%":       null_pct,
        "unique_vals":  df.nunique(),
        "sample_value": [df[c].dropna().iloc[0] if df[c].notna().any() else None
                         for c in df.columns],
    })

    print(f"Shape   : {total_rows:,} rows × {total_cols} columns")
    print(f"Memory  : {df.memory_usage(deep=True).sum() / 1e6:.1f} MB")
    print()
    print(summary.to_string())


data_overview(df)
````

````python
# ── Optional: Monthly Invoice Summary ─────────────────────────────────────────
# Adjust column names to match your actual schema
# Example: 'InvoiceDate', 'InvoiceAmount'

DATE_COL   = "InvoiceDate"    # ← your date column
AMOUNT_COL = "InvoiceAmount"  # ← your amount/value column (or None)

if DATE_COL in df.columns and df[DATE_COL].notna().any():
    df["_Year"]  = df[DATE_COL].dt.year
    df["_Month"] = df[DATE_COL].dt.month

    group_cols = ["_Year", "_Month"]

    if AMOUNT_COL in df.columns:
        monthly = (
            df.groupby(group_cols, sort=True)
              .agg(
                  Invoice_Count   = (DATE_COL,   "count"),
                  Total_Amount    = (AMOUNT_COL, "sum"),
                  Avg_Amount      = (AMOUNT_COL, "mean"),
              )
              .reset_index()
              .rename(columns={"_Year": "Year", "_Month": "Month"})
        )
    else:
        monthly = (
            df.groupby(group_cols, sort=True)
              .size()
              .reset_index(name="Invoice_Count")
              .rename(columns={"_Year": "Year", "_Month": "Month"})
        )

    print("Monthly Invoice Summary:")
    monthly
else:
    print(f"⚠️  Column '{DATE_COL}' not found or all NaT — skipping monthly summary.")
````

## 🔧 Quick Reference — Date Parsing Cheatsheet

```python
# Basic conversion (most common fix)
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], errors='coerce')

# If dates are stored as strings like '25/01/2024' (day-first)
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], dayfirst=True, errors='coerce')

# If dates are epoch milliseconds (Unix timestamps)
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], unit='ms', errors='coerce')

# If SQL Server sends timezone-aware datetimes
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], utc=True, errors='coerce').dt.tz_convert(None)

# Format the date for display
df['InvoiceDate_str'] = df['InvoiceDate'].dt.strftime('%d %b %Y')  # e.g. '15 Jan 2024'
```
