## 2. DataFrame Clean needed

```python
import pandas as pd

# Define null-like values
null_values = [0, '0', '-', 'null', None]

all_null_string_columns = []

for column in df.columns:
    # Normalize values (convert to string + strip spaces + lower)
    col_series = df[column].astype(str).str.strip().str.lower()

    # Check if all values belong to null_values
    if col_series.isin([str(x).lower() for x in null_values]).all():
        all_null_string_columns.append(column)

print(f"Number of columns where all values are null-like: {len(all_null_string_columns)}")

if all_null_string_columns:
    print("\nColumns with only null-like values:")
    for col in all_null_string_columns:
        print(f"- {col}")
else:
    print("No columns found where all values are null-like.")
```

```python
# ----------------------------
# Column Definitions
# ----------------------------
date_columns = [
    "BL Date", "Bank Submit Date", "EXP Date", "Ex Factory",
    "ExfactoryDate", "Expected Realize Date", "FCR Date",
    "Invoice Date", "On Board Date", "Payment Due Date",
    "Realize Date", "Shipping Bill Date"
]

numeric_int_columns = [
    "Commission", "Invoice Qty", "LC/SC Quantity"
]

numeric_float_columns = [
    "Discount", "Invoice Value", "LC/SC Value"
]

# ----------------------------
# Helper: Safe Date Parser
# ----------------------------
def safe_to_datetime(series):
    # Step 1: Clean text
    s = series.astype(str).str.strip()

    # Replace common junk values
    s = s.replace(
        ["", "nan", "None", "-", "--", "null", "NaT"],
        np.nan
    )

    # Step 2: Try multiple parsing strategies
    parsed = pd.to_datetime(s, errors='coerce', dayfirst=True)

    # Step 3: Try fallback (different format)
    fallback = pd.to_datetime(s, errors='coerce', format='%Y-%m-%d')

    # Step 4: Combine (priority: parsed → fallback → original)
    final = parsed.fillna(fallback)

    return final

# ----------------------------
# Helper: Safe Numeric Parser
# ----------------------------
def safe_to_numeric(series, is_int=False):
    s = series.astype(str).str.strip()

    # Remove commas, currency, spaces
    s = s.str.replace(',', '', regex=False)
    s = s.str.replace('৳', '', regex=False)
    s = s.str.replace('$', '', regex=False)

    # Replace junk
    s = s.replace(
        ["", "nan", "None", "-", "--", "null"],
        np.nan
    )

    num = pd.to_numeric(s, errors='coerce')

    if is_int:
        return num.round().astype("Int64")

    return num

# ----------------------------
# Convert DATE columns (SAFE)
# ----------------------------
for col in date_columns:
    if col in df.columns:
        original = df[col].copy()

        converted = safe_to_datetime(df[col])

        # Preserve original where conversion failed
        df[col] = converted.where(converted.notna(), original)

# ----------------------------
# Convert INTEGER columns (SAFE)
# ----------------------------
for col in numeric_int_columns:
    if col in df.columns:
        original = df[col].copy()

        converted = safe_to_numeric(df[col], is_int=True)

        df[col] = converted.where(converted.notna(), original)

# ----------------------------
# Convert FLOAT columns (SAFE)
# ----------------------------
for col in numeric_float_columns:
    if col in df.columns:
        original = df[col].copy()

        converted = safe_to_numeric(df[col], is_int=False)

        df[col] = converted.where(converted.notna(), original)

# ----------------------------
# Clean TEXT columns
# ----------------------------
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

    df[col] = df[col].replace(
        ["nan", "None", "null", ""],
        np.nan
    )

# ----------------------------
# Optional: Debug problematic values
# ----------------------------
def find_bad_dates(df, col):
    mask = pd.to_datetime(df[col], errors='coerce').isna() & df[col].notna()
    return df.loc[mask, col].unique()

# Example debug:
# print(find_bad_dates(df, "Invoice Date"))

# ----------------------------
# Final Info
# ----------------------------
df.info()
```

```python
cols = df.columns.tolist()

if 'PaymentAfterExFactory' in cols and 'Invoice_No' in cols:
    cols.insert(cols.index('Invoice_No'), cols.pop(cols.index('PaymentAfterExFactory')))
    df = df[cols]
else:
    raise KeyError("Missing required columns for reorder")
```

```python
import pandas as pd
import numpy as np

# =========================================================
# REALIZE PENDING REPORT
# =========================================================

Realize_Report = df[df['Invoice_Status'] == "Realize Pending"].copy()

# =========================================================
# NORMALIZE COLUMNS
# =========================================================

Realize_Report.columns = Realize_Report.columns.str.strip()

Realize_Report = Realize_Report.loc[
    :,
    ~Realize_Report.columns.duplicated()
]

# =========================================================
# DATE CONVERSION
# =========================================================

date_cols = [
    'Invoice_Date',
    'EXP_Date',
    'Ex_Factory_Date',
    'On_Board_Date',
    'Bank_Submit_Date',
    'Realize_Date'
]

for col in date_cols:
    if col in Realize_Report.columns:
        Realize_Report[col] = pd.to_datetime(
            Realize_Report[col],
            errors='coerce'
        )

# =========================================================
# NUMERIC CONVERSION
# NOTE: LCSC_Payment_Term is intentionally excluded
# because it contains mixed text and numeric values.
# =========================================================

numeric_cols = [
    'Invoice_Value',
    'PaymentAfterExFactory'
]

for col in numeric_cols:
    if col in Realize_Report.columns:
        Realize_Report[col] = pd.to_numeric(
            Realize_Report[col],
            errors='coerce'
        )

# =========================================================
# CLEAN LCSC PAYMENT TERM AS TEXT
# =========================================================

if 'LCSC_Payment_Term' in Realize_Report.columns:
    Realize_Report['LCSC_Payment_Term'] = (
        Realize_Report['LCSC_Payment_Term']
        .astype('string')
        .str.strip()
    )

    Realize_Report.loc[
        Realize_Report['LCSC_Payment_Term']
        .str.lower()
        .isin(['nan', 'none', 'nat', '']),
        'LCSC_Payment_Term'
    ] = pd.NA

# =========================================================
# NORMALIZE BANK REF NO
# =========================================================

if 'FDBC_No' in Realize_Report.columns:
    Realize_Report['FDBC_No'] = (
        Realize_Report['FDBC_No']
        .astype(str)
        .str.strip()
    )

    Realize_Report.loc[
        Realize_Report['FDBC_No']
        .str.lower()
        .isin(['nan', 'none', '']),
        'FDBC_No'
    ] = np.nan

Realize_Report = Realize_Report.dropna(
    subset=['FDBC_No']
).copy()

# =========================================================
# DETECT INVOICE COLUMN
# =========================================================

invoice_col = 'Invoice_No'

if invoice_col not in Realize_Report.columns:
    if 'Invoice' in Realize_Report.columns:
        invoice_col = 'Invoice'
    else:
        raise ValueError("Invoice column not found")

# =========================================================
# HELPER: FIRST VALID VALUE
# =========================================================

def first_valid(series):
    series = series.dropna()

    if series.empty:
        return ""

    return series.iloc[0]

# =========================================================
# ONBOARD DATE RANGE
# =========================================================

onboard_range = (
    Realize_Report
    .groupby('FDBC_No', as_index=False)['On_Board_Date']
    .agg(
        min_onboard='min',
        max_onboard='max'
    )
)

# =========================================================
# UNIQUE INVOICE LIST
# =========================================================

invoice_list = (
    Realize_Report
    .dropna(subset=[invoice_col])
    .assign(
        __invoice=lambda d:
        d[invoice_col]
        .astype(str)
        .str.strip()
    )
    .sort_values('__invoice')
    .drop_duplicates(['FDBC_No', '__invoice'])
    .groupby('FDBC_No', as_index=False)['__invoice']
    .agg(
        INVOICES=lambda s: ", ".join(s)
    )
)

# =========================================================
# MAIN SUMMARY
# =========================================================

summary = (
    Realize_Report
    .groupby('FDBC_No', as_index=False)
    .agg(
        Status=('Invoice_Status', first_valid),
        Category=('Category', first_valid),
        Buyer=('Buyer', first_valid),
        Company=('LC_Factory', first_valid),
        Bank=('Bank_Name', first_valid),

        PaymentTermXFactory=(
            'PaymentAfterExFactory',
            first_valid
        ),

        LCSC_Payment_Term=(
            'LCSC_Payment_Term',
            first_valid
        ),

        ExFactoryDate=(
            'Ex_Factory_Date',
            'max'
        ),

        BankSubmitDate=(
            'Bank_Submit_Date',
            'max'
        ),

        InvoiceDate=(
            'Invoice_Date',
            'max'
        ),

        Total_Invoices=(
            invoice_col,
            lambda x: x.nunique()
        ),

        Total_Invoice_Value=(
            'Invoice_Value',
            'sum'
        )
    )
)

# =========================================================
# MERGE EXTRA DATA
# =========================================================

summary = summary.merge(
    onboard_range,
    on='FDBC_No',
    how='left'
)

summary = summary.merge(
    invoice_list,
    on='FDBC_No',
    how='left'
)

if 'INVOICES' not in summary.columns:
    summary['INVOICES'] = ""

# =========================================================
# BANK SUBMIT STATUS
# =========================================================

summary['IsBankSubmit?'] = np.where(
    summary['BankSubmitDate'].notna(),
    'YES',
    'NO'
)

# =========================================================
# ONBOARD DATE RANGE FORMAT
# =========================================================

def format_onboard(rmin, rmax):
    if pd.isna(rmin) or pd.isna(rmax):
        return ""

    return (
        f"{rmin.strftime('%d-%b-%Y')} "
        f"to "
        f"{rmax.strftime('%d-%b-%Y')}"
    )

summary['OnBoardDate(Range)'] = summary.apply(
    lambda row: format_onboard(
        row['min_onboard'],
        row['max_onboard']
    ),
    axis=1
)

summary.drop(
    columns=['min_onboard', 'max_onboard'],
    inplace=True
)

# =========================================================
# REALIZABLE STATUS
# =========================================================

today = pd.Timestamp.today().normalize()

summary['PaymentTermXFactory'] = pd.to_numeric(
    summary['PaymentTermXFactory'],
    errors='coerce'
)

summary['Due_Date'] = pd.NaT

mask = (
    summary['PaymentTermXFactory'].notna()
    &
    summary['ExFactoryDate'].notna()
)

summary.loc[mask, 'Due_Date'] = (
    summary.loc[mask, 'ExFactoryDate']
    +
    pd.to_timedelta(
        summary.loc[mask, 'PaymentTermXFactory'],
        unit='D'
    )
)

def realizable_status(row):
    if pd.isna(row['Due_Date']):
        return ""

    if today > row['Due_Date']:
        return "NeedRealize"

    return "NotYetRealize"

summary['IsFDBCRealizable?'] = summary.apply(
    realizable_status,
    axis=1
)

summary.drop(
    columns=['Due_Date'],
    inplace=True
)

# =========================================================
# SORT DATA
# =========================================================

summary = summary.sort_values(
    by='ExFactoryDate',
    ascending=True
)

# =========================================================
# SERIAL NUMBER
# =========================================================

summary.insert(
    0,
    'SL#',
    range(1, len(summary) + 1)
)

# =========================================================
# FINAL COLUMN ORDER
# =========================================================

final_cols = [
    'SL#',
    'FDBC_No',
    'Status',
    'Category',
    'Buyer',
    'Company',
    'IsBankSubmit?',
    'ExFactoryDate',
    'Bank',
    'BankSubmitDate',
    'PaymentTermXFactory',
    'LCSC_Payment_Term',
    'OnBoardDate(Range)',
    'Total_Invoices',
    'Total_Invoice_Value',
    'InvoiceDate',
    'IsFDBCRealizable?',
    'INVOICES'
]

for col in final_cols:
    if col not in summary.columns:
        summary[col] = ""

summary = summary[final_cols]

# =========================================================
# FINAL CLEANING
# =========================================================

summary['Total_Invoice_Value'] = (
    pd.to_numeric(
        summary['Total_Invoice_Value'],
        errors='coerce'
    )
    .fillna(0)
    .round(2)
)

summary['LCSC_Payment_Term'] = (
    summary['LCSC_Payment_Term']
    .fillna("")
    .astype(str)
    .str.strip()
)

# =========================================================
# RENAME COLUMNS
# =========================================================

summary.rename(
    columns={
        'FDBC_No': 'Bank Ref No',
        'PaymentTermXFactory': 'Payment Term (X-Factory)',
        'Total_Invoice_Value': 'Total Invoice Value'
    },
    inplace=True
)

# =========================================================
# FINAL OUTPUT
# =========================================================

RealizePending = summary.copy()

print("Realize Pending Table Created Successfully")
```
