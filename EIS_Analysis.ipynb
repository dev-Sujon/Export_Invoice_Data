## 2. DataFrame Clean needed

```python
import pandas as pd

# Define null-like values
null_values = [0, '0', '-', 'null', None]

all_null_string_columns = []

for column in df.columns:
    # Normalize values (convert to string + strip spaces + lower)
    col_series = df[column].astype(str).str.strip().str.lower()

    # Check if all values belong to null_values
    if col_series.isin([str(x).lower() for x in null_values]).all():
        all_null_string_columns.append(column)

print(f"Number of columns where all values are null-like: {len(all_null_string_columns)}")

if all_null_string_columns:
    print("\nColumns with only null-like values:")
    for col in all_null_string_columns:
        print(f"- {col}")
else:
    print("No columns found where all values are null-like.")
```

```python
# ----------------------------
# Column Definitions
# ----------------------------
date_columns = [
    "BL Date", "Bank Submit Date", "EXP Date", "Ex Factory",
    "ExfactoryDate", "Expected Realize Date", "FCR Date",
    "Invoice Date", "On Board Date", "Payment Due Date",
    "Realize Date", "Shipping Bill Date"
]

numeric_int_columns = [
    "Commission", "Invoice Qty", "LC/SC Quantity"
]

numeric_float_columns = [
    "Discount", "Invoice Value", "LC/SC Value"
]

# ----------------------------
# Helper: Safe Date Parser
# ----------------------------
def safe_to_datetime(series):
    # Step 1: Clean text
    s = series.astype(str).str.strip()

    # Replace common junk values
    s = s.replace(
        ["", "nan", "None", "-", "--", "null", "NaT"],
        np.nan
    )

    # Step 2: Try multiple parsing strategies
    parsed = pd.to_datetime(s, errors='coerce', dayfirst=True)

    # Step 3: Try fallback (different format)
    fallback = pd.to_datetime(s, errors='coerce', format='%Y-%m-%d')

    # Step 4: Combine (priority: parsed → fallback → original)
    final = parsed.fillna(fallback)

    return final

# ----------------------------
# Helper: Safe Numeric Parser
# ----------------------------
def safe_to_numeric(series, is_int=False):
    s = series.astype(str).str.strip()

    # Remove commas, currency, spaces
    s = s.str.replace(',', '', regex=False)
    s = s.str.replace('৳', '', regex=False)
    s = s.str.replace('$', '', regex=False)

    # Replace junk
    s = s.replace(
        ["", "nan", "None", "-", "--", "null"],
        np.nan
    )

    num = pd.to_numeric(s, errors='coerce')

    if is_int:
        return num.round().astype("Int64")

    return num


# ----------------------------
# Convert DATE columns (SAFE)
# ----------------------------
for col in date_columns:
    if col in df.columns:
        original = df[col].copy()

        converted = safe_to_datetime(df[col])

        # Preserve original where conversion failed
        df[col] = converted.where(converted.notna(), original)

# ----------------------------
# Convert INTEGER columns (SAFE)
# ----------------------------
for col in numeric_int_columns:
    if col in df.columns:
        original = df[col].copy()

        converted = safe_to_numeric(df[col], is_int=True)

        df[col] = converted.where(converted.notna(), original)

# ----------------------------
# Convert FLOAT columns (SAFE)
# ----------------------------
for col in numeric_float_columns:
    if col in df.columns:
        original = df[col].copy()

        converted = safe_to_numeric(df[col], is_int=False)

        df[col] = converted.where(converted.notna(), original)

# ----------------------------
# Clean TEXT columns
# ----------------------------
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

    df[col] = df[col].replace(
        ["nan", "None", "null", ""],
        np.nan
    )

# ----------------------------
# Optional: Debug problematic values
# ----------------------------
def find_bad_dates(df, col):
    mask = pd.to_datetime(df[col], errors='coerce').isna() & df[col].notna()
    return df.loc[mask, col].unique()

# Example debug:
# print(find_bad_dates(df, "Invoice Date"))

# ----------------------------
# Final Info
# ----------------------------
df.info()
```

```python
# Merge based on Buyer name
df = df.merge(
    Buyer_Payment[['Buyer', 'Payment After Ex-Factory']],
    on='Buyer',
    how='left'   # Keep all EIS rows
)
# Rename column
df.rename(columns={
    'Payment After Ex-Factory': 'PaymentAfterExFactory'
}, inplace=True)
```

```python
cols = df.columns.tolist()

if 'PaymentAfterExFactory' in cols and 'Invoice_No' in cols:
    cols.insert(cols.index('Invoice_No'), cols.pop(cols.index('PaymentAfterExFactory')))
    df = df[cols]
else:
    raise KeyError("Missing required columns for reorder")
```

```python
Realize_Report = df[df['Invoice_Status'] == "Realize Pending"]
```

```python
Realize_Report.columns
```

```python
# -------------------------------
# INPUT DATAFRAME (change name if needed)
# -------------------------------
Realize_Report = df.copy()

# -------------------------------
# NORMALIZE COLUMNS & DROP DUPLICATES
# -------------------------------
Realize_Report.columns = Realize_Report.columns.str.strip()
Realize_Report = Realize_Report.loc[:, ~Realize_Report.columns.duplicated()]

# -------------------------------
# SAFE DATE & NUMERIC CONVERSIONS
# -------------------------------
date_cols = [
    'Invoice_Date', 'EXP_Date', 'Ex_Factory_Date',
    'On_Board_Date', 'Bank_Submit_Date', 'Realize_Date'
]
for c in date_cols:
    if c in Realize_Report.columns:
        Realize_Report[c] = pd.to_datetime(Realize_Report[c], errors='coerce')

numeric_cols = ['Invoice_Value', 'PaymentAfterExFactory']
for c in numeric_cols:
    if c in Realize_Report.columns:
        Realize_Report[c] = pd.to_numeric(Realize_Report[c], errors='coerce')

# -------------------------------
# NORMALIZE/VALIDATE BANK REF (FDBC_No)
# -------------------------------
if 'FDBC_No' in Realize_Report.columns:
    Realize_Report['FDBC_No'] = Realize_Report['FDBC_No'].astype(str).str.strip()
    Realize_Report.loc[Realize_Report['FDBC_No'].str.lower() == 'nan', 'FDBC_No'] = np.nan

# drop rows without a Bank Ref No
Realize_Report = Realize_Report.dropna(subset=['FDBC_No']).copy()

# -------------------------------
# PREP: invoice column
# -------------------------------
invoice_col = 'Invoice_No'
if invoice_col not in Realize_Report.columns and 'Invoice' in Realize_Report.columns:
    invoice_col = 'Invoice'

# -------------------------------
# ONBOARD DATE RANGE (group-level)
# -------------------------------
onboard_range = (
    Realize_Report.groupby('FDBC_No', as_index=False)['On_Board_Date']
      .agg(min_onboard='min', max_onboard='max')
)

# -------------------------------
# INVOICE LIST (unique, sorted)
# -------------------------------
invoice_list = (
    Realize_Report.dropna(subset=[invoice_col])
      .assign(__inv_str=lambda d: d[invoice_col].astype(str).str.strip())
      .sort_values('__inv_str')
      .drop_duplicates(['FDBC_No', '__inv_str'])
      .groupby('FDBC_No', as_index=False)['__inv_str']
      .agg(INVOICES=lambda s: ",".join(s))
)

# -------------------------------
# AGGREGATE / GROUP BY FDBC_No
# -------------------------------
summary = (
    Realize_Report.groupby('FDBC_No', as_index=False)
      .agg(
          Status=('Invoice_Status', 'first'),
          Category=('Category', 'first'),
          Buyer=('Buyer', 'first'),
          Company=('LC_Factory', 'first'),           # Factory -> Company
          Bank=('Bank_Name', 'first'),
          PaymentTermXFactory=('PaymentAfterExFactory', 'first'),
          ExFactoryDate=('Ex_Factory_Date', 'max'),    # last date
          BankSubmitDate=('Bank_Submit_Date', 'max'),  # last date
          InvoiceDate=('Invoice_Date', 'max'),        # last invoice date
          Total_Invoices=(invoice_col, lambda s: s.dropna().count()),
          Total_Invoice_Value=('Invoice_Value', 'sum')
      )
)

# -------------------------------
# MERGE onboard_range & invoice_list
# -------------------------------
summary = summary.merge(onboard_range, on='FDBC_No', how='left')
summary = summary.merge(invoice_list, on='FDBC_No', how='left')

# Ensure INVOICES column exists
if 'INVOICES' not in summary.columns:
    summary['INVOICES'] = ""

# -------------------------------
# IsBankSubmit? (YES/NO)
# -------------------------------
summary['IsBankSubmit?'] = np.where(summary['BankSubmitDate'].notna(), 'YES', 'NO')

# -------------------------------
# OnBoardDate(Range) formatting
# -------------------------------
def format_onboard(rmin, rmax):
    if pd.isna(rmin) or pd.isna(rmax):
        return ""
    return f"{rmin.strftime('%d-%b-%Y')} to {rmax.strftime('%d-%b-%Y')}"

summary['OnBoardDate(Range)'] = summary.apply(
    lambda r: format_onboard(r['min_onboard'], r['max_onboard']), axis=1
)

summary.drop(columns=['min_onboard', 'max_onboard'], inplace=True)

# -------------------------------
# IsFDBCRealizable? (group-level)
# -------------------------------
today = pd.Timestamp.today().normalize()
summary['PaymentTermXFactory'] = pd.to_numeric(summary['PaymentTermXFactory'], errors='coerce')

summary['Due_Date'] = pd.NaT
mask = summary['PaymentTermXFactory'].notna() & summary['ExFactoryDate'].notna()
if mask.any():
    summary.loc[mask, 'Due_Date'] = (
        summary.loc[mask, 'ExFactoryDate'] +
        pd.to_timedelta(summary.loc[mask, 'PaymentTermXFactory'].astype(float), unit='D')
    )

def realizable_status(row):
    if pd.isna(row['PaymentTermXFactory']) or pd.isna(row['Due_Date']):
        return ""
    return 'NeedRealize' if today > row['Due_Date'] else 'NotYetRealize'

summary['IsFDBCRealizable?'] = summary.apply(realizable_status, axis=1)
summary.drop(columns=['Due_Date'], inplace=True)

# -------------------------------
# FINAL CLEANUP & ORDERING
# -------------------------------
summary.insert(0, 'SL#', range(1, len(summary) + 1))

final_cols = [
    'SL#',
    'FDBC_No',                # will rename to 'Bank Ref No'
    'Status',
    'Category',
    'Buyer',
    'Company',
    'IsBankSubmit?',
    'ExFactoryDate',
    'Bank',
    'BankSubmitDate',
    'PaymentTermXFactory',
    'OnBoardDate(Range)',
    'Total_Invoices',
    'Total_Invoice_Value',
    'InvoiceDate',            # last Invoice date added
    'IsFDBCRealizable?',
    'INVOICES'
]

for c in final_cols:
    if c not in summary.columns:
        summary[c] = "" if summary.shape[0] > 0 else pd.Series(dtype=object)

summary = summary[final_cols]

summary['Total_Invoice_Value'] = pd.to_numeric(summary['Total_Invoice_Value'], errors='coerce').fillna(0)

# -------------------------------
# FINAL OUTPUT
# -------------------------------
RealizePending = summary.copy()
print("Realize Pending Table Created Successfully")
```
